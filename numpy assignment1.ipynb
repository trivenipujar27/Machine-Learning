```python
import numpy as np

prices = np.array([120, 250, 80, 450, 300])
quantities = np.array([4, 2, 10, 1, 3])

revenue = prices * quantities
total_revenue = np.sum(revenue)
average_revenue = np.mean(revenue)
highest_index = np.argmax(revenue)

print("Revenue generated by each product:", revenue)
print("Overall combined revenue: ₹", total_revenue)
print("Average revenue per product: ₹", average_revenue)
print("Index of highest-earning product:", highest_index)
print("Highest revenue: ₹", revenue[highest_index])
```

```python
import numpy as np

rainfall = np.array([12.5, -1.0, 0.0, 18.2, -99.0,
                     5.4, 22.1, 0.0, -1.0, 14.8])

valid_rainfall = rainfall[rainfall >= 0]
print("Valid rainfall readings:", valid_rainfall)

rainfall = np.where(rainfall < 0, 0.0, rainfall)
print("Cleaned rainfall array:", rainfall)

heavy_days = np.count_nonzero(rainfall > 10.0)
print("Number of heavy rainfall days:", heavy_days)
```

```python
import numpy as np
np.random.seed(10)

scores = np.random.randint(20, 101, 12)

print("Original scores:", scores)

status = np.where(scores >= 40, "Pass", "Fail")
print("Student status:", status)

updated_scores = np.where(scores < 40, scores + 5, scores)
print("Updated scores:", updated_scores)
```

```python
import numpy as np
marks = np.random.randint(50, 101, (6, 4))

print("Marks Matrix:")
print(marks)

student_avg = np.round(np.mean(marks, axis=1), 2)
print("\nAverage marks of each student:")
print(student_avg)

subject_max = np.max(marks, axis=0)
print("\nHighest marks in each subject:")
print(subject_max)

filtered_students = marks[student_avg > 75]
print("\nStudents with average marks greater than 75:")
print(filtered_students)
```

```python
import numpy as np

morning_sales = np.array([
    [15, 20],
    [10, 12],
    [25, 30]
])

evening_sales = np.array([
    [18, 22],
    [14, 15],
    [20, 28]
])

full_day_sales = np.hstack((morning_sales, evening_sales))

print("Full day sales (3 × 4):")
print(full_day_sales)

fourth_branch = np.array([[12, 16, 15, 19]])

final_sales = np.vstack((full_day_sales, fourth_branch))

print("\nFinal sales matrix (4 × 4):")
print(final_sales)

column_totals = np.sum(final_sales, axis=0)

print("\nTotal units sold in each column:")
print(column_totals)
```

```python
import numpy as np

emp_ids = np.array(["E101", "E102", "E103", "E104", "E105", "E106"])
scores = np.array([78, 92, 65, 88, 95, 71])

sorted_indices = np.argsort(scores)[::-1]

sorted_emp_ids = emp_ids[sorted_indices]
sorted_scores = scores[sorted_indices]

print("Top 3 Employees:")

for i in range(3):
    print(f"Employee ID: {sorted_emp_ids[i]}, Score: {sorted_scores[i]}")
```
