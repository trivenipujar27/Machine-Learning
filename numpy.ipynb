```python
a=np.array(1,2,3,4)
b=np.array(5,6,7,8)
c=np.stack((a,b),axis=1)
print(c)
```

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
c=np.hstack((a,b))
print(c)
```

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
c=np.vstack((a,b))
print(c)
```

```python
a=np.array([1,2,3,4])
b=a.copy()
a[0]=10
print(a)
print(b)
```

```python
a=np.array([1,2,3,4])
b=a.view()
a[0]=10
print(a)
print(b)
```
